### CREATE A FUNCTION TO PERFORM BASIC DATA CLEANING FOR THE HR DATAFRAME

```python
#Function definition
def wrangle(df):
    #Remove all trailing spaces in the 'Name' column
    df['Name'] = df['Name'].str.strip()
    
    #Replace 'thirty' with '30' in the 'Age' column
    df['Age'] = df['Age'].replace({'thirty':'30'})
    
    #Replace 'SIXTY THOUSAND' with '60000' in the 'Salary' column
    df['Salary'] = df['Salary'].replace({'SIXTY THOUSAND':'60000'})

    #Unify the date format in 'Joining Date' column
    df['Joining Date'] = pd.to_datetime(df['Joining Date'], format='mixed')
    df['Joining Date'] = df['Joining Date'].dt.strftime('%Y-%m-%d')
    
    #Convert 'Age' and 'Salary' columns data type from 'object' to 'int'
    df[['Age', 'Salary']] = df[['Age', 'Salary']].replace({' NAN ': '0'})
    df[['Age', 'Salary']] = df[['Age', 'Salary']].fillna('0')
    df[['Age', 'Salary']] = df[['Age', 'Salary']].astype('int')
    
    return df

```

```python
#Use the wrangle function to clean the hr dataframe
wrangle(hr)
hr.head()
```

#### HANDLE MISSING VALUES IN THE AGE COLUMN

```python
#Get a count of values in the Age column
hr.Age.value_counts()
```

```python
#Get indices of zero(missing) values
zero_indices = hr[hr['Age'] == 0].index

#Shuffle the indices
shuffled_indices = np.random.permutation(zero_indices)

#Calculate allocation counts
n_total = len(shuffled_indices)

n_each = int(n_total * 0.2)
n_30 = n_each
n_25 = n_each
n_40 = n_each
n_35 = n_each
n_50 = n_total - (n_30 + n_25 + n_40 + n_35)

#Assign values to shuffled indices
hr.loc[shuffled_indices[:n_30], 'Age'] = 30
hr.loc[shuffled_indices[n_30:n_30 + n_25], 'Age'] = 25
hr.loc[shuffled_indices[n_30 + n_25:n_30 + n_25 + n_40], 'Age'] = 40
hr.loc[shuffled_indices[n_30 + n_25 + n_40:n_30 + n_25 + n_40 + n_35], 'Age'] = 35
hr.loc[shuffled_indices[n_30 + n_25 + n_40 + n_35:], 'Age'] = 50

```
